```python
import pandas as pd   
```

```python
df=pd.read_csv('Churn_Modelling.csv')
df.head()
```

```python
df.columns
```

```python
df.drop(['RowNumber', 'CustomerId', 'Surname'],axis=1,inplace=True)
```

```python
df
```

```python
from sklearn.preprocessing import LabelEncoder   

label_gender=LabelEncoder()  
df['Gender']=label_gender.fit_transform(df['Gender'])  
```

```python
df['Gender']
```

```python
df.info()
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
one_hot_geo=OneHotEncoder() 
geo_encoded=one_hot_geo.fit_transform(df[['Geography']])
geo_encoded
```

```python
geo_encoded.toarray()
```

```python
one_hot_geo.get_feature_names_out()
```

```python
df_geo=pd.DataFrame(geo_encoded.toarray(),columns=one_hot_geo.get_feature_names_out(['Geography']))
df_geo
```

```python
df
```

```python
df.drop(['Geography'],axis=1,inplace=True)
```

```python
dfn=pd.concat([df,df_geo],axis=1)
```

```python
dfn
```

```python
import pickle as pkl  
with open('label_gender.pkl','wb') as file: 
    pkl.dump(label_gender,file)
```

```python
with open('one_hot_geo.pkl','wb') as file : 
    pkl.dump(one_hot_geo,file)
```

```python
dfn.head()
```

```python
from sklearn.preprocessing import StandardScaler   
from sklearn.model_selection import train_test_split  
x=dfn.drop(['Exited'],axis=1) 
y=dfn['Exited'] 
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42) 
x_train.shape,y_train.shape
```

```python
sc=StandardScaler()
x_train=sc.fit_transform(x_train) 
x_test=sc.transform(x_test)
```

```python
with open("Scaler.pkl",'wb') as file: 
    pkl.dump(sc,file)
```

```python
#ANN 

import tensorflow 

```

```python
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping, TensorBoard

from tensorflow.keras.models import Sequential
```

```python
model=Sequential([
    Dense(32,activation='relu',input_shape=(x_train.shape[1],)),
    Dense(16,activation='relu'),
    Dense(1,activation='sigmoid')])
```

```python
opt=tensorflow.keras.optimizers.Adam(learning_rate=0.001)
loss=tensorflow.keras.losses.BinaryCrossentropy()
model.compile(optimizer=opt,loss=loss,metrics=['accuracy'])
```

```python
#tensor board is visualization tool to visualize the all mention function 
```

```python
import datetime
```

```python
from tensorflow.keras.callbacks import TensorBoard,EarlyStopping
```

```python
log_dir='log/fit/' +datetime.datetime.now().strftime('%y%m%d-%H%M%S')
```

```python
tensorflow_callback=TensorBoard(log_dir,histogram_freq=1)
```

```python
early_stopping_callbacks=EarlyStopping(monitor='val_loss',patience=5,restore_best_weights=True)
```

```python
history=model.fit(x_train,y_train,epochs=100,batch_size=32,validation_data=(x_test,y_test),callbacks=[tensorflow_callback,early_stopping_callbacks])
```

```python
model.save("churn_model.h5")
```
