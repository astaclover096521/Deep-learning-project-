```python
import pandas as pd
import numpy as np 
import pickle as p
import tensorflow as tf
from tensorflow.keras.models import load_model
```

```python
model=load_model('churn_model.h5')
```

```python
with open('label_gender.pkl','rb') as file:
    label_encoder_gender=p.load(file)
```

```python
with open('one_hot_geo.pkl','rb') as file:
    one_hot_geo=p.load(file)
```

```python
with open('Scaler.pkl','rb') as file:
    Scaler=p.load(file)
```

```python
#  Sample data 
input_data = {

'CreditScore':600,

'Geography': 'France',

'Gender': 'Male',

'Age': 40,

'Tenure': 3,

'Balance':60000,

'NumOfProducts': 2,

'HasCrCard':1,

'IsActiveMember': 1,

'EstimatedSalary': 50000

}
```

```python
geo_encoder=one_hot_geo.transform([[input_data['Geography']]]).toarray()
```

```python
geo_encoder
```

```python
geo_df=pd.DataFrame(geo_encoder,columns=one_hot_geo.get_feature_names_out(['Geography']))
```

```python
geo_df
```

```python
input_df=pd.DataFrame([input_data])
input_df
```

```python
input_df['Gender']=label_encoder_gender.transform([input_data['Gender']])
input_df
```

```python
input_df.drop('Geography',axis=1,inplace=True)
```

```python
df=pd.concat([input_df,geo_df],axis=1)
df
```

```python
input_scaled=Scaler.transform(df)
input_scaled
```

```python
pred=model.predict(input_scaled)
pred
```

```python
pred[0][0]    # 0 leave, 1 stay
```

```python
if pred[0][0]>0.5:
    print('Customer will leave')
else:
    print('Customer will stay')
```
